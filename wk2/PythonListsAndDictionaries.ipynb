```python
# Exercise 1: Lists and Dictionaries

"""
Instructions:
1. Create a shopping list with at least 5 items
2. Print out the third item in the list
3. Replace the first item with a different item
4. Add an item to the end of the list
5. Create a dictionary of 3 grocery items with their prices
6. Print a message showing one item and its price
7. Add a new item to the dictionary with its price
8. Calculate and print the total cost of all items
"""
```

```python
# Your solution here:   # prof says it's simple so students work independently.
# Create shopping list
shopping_list = ["watermelon", "milk", "bread", "eggs", "coffee"]
```

```python
# Print third item
print(shopping_list[2])
```

```python
# Replace first item
shopping_list[0] = "blueberries"
print(shopping_list)
```

```python
# Add item to end
shopping_list.append("cheese")

print(shopping_list)
```

```python
# Create grocery dictionary
grocery_prices = {
    "watermelon": 4.50,
    "milk": 6.50,
    "bread": 2.50
}
```

```python
# Print item and price
print("The price of milk is $", grocery_prices["milk"])
```

```python
# Add new item
grocery_prices["eggs"] = 2.80
```

```python
# Calculate total cost
total_cost = sum(grocery_prices.values())

print("Total cost: $", total_cost)
```
