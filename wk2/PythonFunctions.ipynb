```python
# Task 7: Combining Functions and Conditionals
# Create a function named 'analyze_temperature' that:
# 1. Takes a temperature value and a unit ('C' or 'F')
# 2. Converts the temperature to the other unit
# 3. Returns a string describing if it's "freezing" (below 0°C), 
#    "cold" (0-15°C), "moderate" (15-25°C), or "hot" (above 25°C)
# Your code here


# Test Task 7
print("Task 7 Test:")
# Test with different temperatures in both Celsius and Fahrenheit
# Print the results
print()
```

```python
# prof says helper functions from Task 6
def celsius_to_fahrenheit(celsius): 
    return celsius * 9/5 + 32
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9
    
def analyze_temperature(temp, unit):   # parameters are temp & unit ("C" or "F")
    # Convert to the other unit
    if unit == "C":
        converted = (temp * 9/5) + 32   # Celsius → Fahrenheit
        celsius = temp
        other_unit = "F"
    elif unit == "F":
        converted = (temp - 32) * 5/9   # Fahrenheit → Celsius
        celsius = converted
        other_unit = "C"
    else:
        return "Invalid unit. Please use 'C' or 'F'."

    # Determine temperature description using Celsius
    if celsius < 0:
        description = "freezing"
    elif celsius <= 15:
        description = "cold"
    elif celsius <= 25:
        description = "moderate"
    else:
        description = "hot"
    return f"{temp}°{unit} = {converted:.1f}°{other_unit}. It is {description}."
```

```python
# Test Task 7
print("Task 7 Test:")
print(analyze_temperature(20, "C"))
print(analyze_temperature(32, "F"))
print(analyze_temperature(-5, "C"))
print(analyze_temperature(95, "F"))
```

```python
# Task 8: Function with Variable Arguments
# Create a function named 'calculate_average' that:
# 1. Accepts any number of numeric arguments   # any number or any data, any argument: prof says Python allows but depends what you want to do
# 2. Returns the average of all numbers provided
# Your code here


# Test Task 8
print("Task 8 Test:")
# Call your calculate_average function with different sets of numbers
# Print the results
print()
```

```python
# *args
# **kwargs
def calculate_average(*numbers):
    if not numbers:
        return 0 

    total = sum(numbers)
    count = len(numbers)
    return total / count
```

```python
# Test Task 8
print("Task 8 Test:")
# Call with different sets of numbers
print(f"Average of (10, 20, 30): {calculate_average(10, 20, 30)}")
print(f"Average of (5, 15): {calculate_average(5, 15)}")
print(f"Average of no numbers: {calculate_average()}")
print()
```

```python
# Task 9: Advanced Challenge - Student Grade Function
# Create a function named 'calculate_final_grade' that:
# 1. Takes a student's name and any number of assignment scores
# 2. Calculates the average score
# 3. Determines the letter grade based on the average (A: 90+, B: 80-89, C: 70-79, D: 60-69, F: <60)
# 4. Returns a formatted string with the student's name, average score, and letter grade
# Your code here


# Test Task 9
print("Task 9 Test:")
# Test with different student names and scores
# Print the results
print()
```

```python
# *args
def calculate_final_grade(name, *scores):   # define a function

    # 1. Handle cases with no scores to avoid division by zero
    if not scores:
        return f"Student {name} has no scores recorded."
    
    # 2. Calculate the average
    average = sum(scores) / len(scores)     # logic
    
    # 3. Determine the letter grade using conditional logic   
    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    else:
        grade = "F"
    return f"{name}: Average = {average:.2f}, Grade = {grade}"   # :.2f displays the average with 2 decimal places.
```

```python
# Test Task 9
print("Task 9 Test:")

print(calculate_final_grade("Alice", 90, 85, 95))
print(calculate_final_grade("Bob", 78, 82, 75))
print(calculate_final_grade("Madison", 80, 99, 75))
print(calculate_final_grade("Charlie", 60, 65, 58))
print(calculate_final_grade("David", 50, 45, 55))
print(calculate_final_grade("Will", 100, 95))
print(calculate_final_grade("Kelly"))
```
